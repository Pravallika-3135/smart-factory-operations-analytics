```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("manufacturing_6G_dataset.csv")
```

```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
df.nunique()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
```

```python
df.info()
```

```python
df["Year"] = df["Timestamp"].dt.year
```

```python
df["Month"] = df["Timestamp"].dt.month_name()
```

```python
df["Day"] = df["Timestamp"].dt.day_name()
```

```python
df["Hour"] = df["Timestamp"].dt.hour
```

```python
df["Shift"] = pd.cut(
    df["Hour"],
    bins=[-1, 7, 15, 23],
    labels=["Night", "Morning", "Evening"]
)
```

```python
df["Energy_Efficiency"] = (
    df["Production_Speed_units_per_hr"] /
    df["Power_Consumption_kW"]
).round(2)
```

```python
df["Maintenance_Risk"] = pd.cut(
    df["Predictive_Maintenance_Score"],
    bins=[0, 40, 70, 100],
    labels=["High", "Medium", "Low"]
)
```

```python
df.head()
```

```python
df.to_csv("Smart_Factory_Cleaned.csv", index=False)
```

```python
daily_prod = df.groupby(df["Timestamp"].dt.date)["Production_Speed_units_per_hr"].sum()

daily_prod.plot(
    kind="line",
    figsize=(12,5),
    marker="o",
    title="Daily Production Trend"
)

plt.xlabel("Date")
plt.ylabel("Production")
plt.show()
```

```python
df.groupby("Shift")["Production_Speed_units_per_hr"]\
.sum()\
.plot(
kind="bar",
figsize=(8,5),
title="Production by Shift"
)

plt.ylabel("Production")
plt.show()
```

```python
df.groupby("Machine_ID")["Production_Speed_units_per_hr"]\
.mean()\
.sort_values(ascending=False)\
.head(10)\
.plot(
kind="bar",
figsize=(10,5),
title="Top 10 Machines by Production"
)

plt.show()
```

```python
df.groupby("Machine_ID")["Temperature_C"]\
.mean()\
.sort_values(ascending=False)\
.head(10)\
.plot(
kind="bar",
figsize=(10,5),
title="Average Temperature by Machine"
)

plt.show()
```

```python
df.groupby("Operation_Mode")["Defect_Rate"]\
.mean()\
.plot(
kind="bar",
figsize=(8,5),
title="Average Defect Rate by Operation Mode"
)

plt.ylabel("Defect Rate")
plt.show()
```

```python
df["Energy_Efficiency"].plot(
kind="hist",
bins=40,
figsize=(8,5),
title="Energy Efficiency Distribution"
)

plt.show()
```

```python
df["Maintenance_Risk"]\
.value_counts()\
.plot(
kind="pie",
autopct="%1.1f%%",
figsize=(6,6),
title="Maintenance Risk Distribution"
)

plt.ylabel("")
plt.show()
```

```python
df.groupby("Shift")["Power_Consumption_kW"]\
.mean()\
.plot(
kind="bar",
figsize=(8,5),
title="Average Power Consumption by Shift"
)

plt.ylabel("Power (kW)")
plt.show()
```

```python
df.groupby("Machine_ID")["Predictive_Maintenance_Score"]\
.mean()\
.sort_values(ascending=False)\
.head(10)\
.plot(
kind="bar",
figsize=(10,5),
title="Top 10 Machines by Maintenance Score"
)

plt.show()
```

```python
corr = df[
[
"Temperature_C",
"Power_Consumption_kW",
"Production_Speed_units_per_hr",
"Defect_Rate",
"Energy_Efficiency"
]
].corr()

plt.figure(figsize=(6,5))

plt.imshow(corr)

plt.xticks(range(len(corr.columns)),corr.columns,rotation=45)
plt.yticks(range(len(corr.columns)),corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j,
            i,
            round(corr.iloc[i,j],2),
            ha="center",
            va="center"
        )

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```
